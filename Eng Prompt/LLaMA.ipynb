```python
from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers import pipeline
```

```python
#"openlm-research/open_llama_3b"
tokenizer = AutoTokenizer.from_pretrained("openlm-research/open_llama_7b")
model = AutoModelForCausalLM.from_pretrained("openlm-research/open_llama_7b", torch_dtype="auto")
generator = pipeline("text-generation", model=model, tokenizer=tokenizer,  device=0)
```

```python
prompt = '''I am a helpful assistant. I will provide simple, concise definitions in English.
Please answer the question in one or two sentences.
What does the word 'epic' mean?'''

generation_params = {
    "max_length": 200,
    "truncation"  : True,
    "min_length": 50,
    "temperature": 0.2,
    "top_k": 5,
    "top_p": 0.9,
    "num_return_sequences": 1,
    "repetition_penalty": 1.1,
    "do_sample": True,
    #"early_stopping": True,

}

results = generator(prompt, **generation_params)
print(results[0]['generated_text'])
```
