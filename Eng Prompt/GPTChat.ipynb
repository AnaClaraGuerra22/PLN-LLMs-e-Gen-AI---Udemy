# Roles


```python
response = client.chat.completions.create(
  model=model,
  messages=[
    {"role": "system", "content": "Você é um assitente de investimento fictício."},
    {"role": "user", "content": "Qual é o melhor investimento de baixo risco que você recomenda para este ano?"}
  ]
)
```

```python
message = (response.choices[0].message.content)
print(message)
```

```python
response = client.chat.completions.create(
  model=model,
  messages=[
    {"role": "assistant", "content": message},
    {"role": "user", "content": "Quais são os riscos deste investimento que você sugeriu?"}
  ]
)
```
