```python
import ipywidgets as widgets
from IPython.display import display, clear_output

from openai import OpenAI

api_key = "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"
client = OpenAI(api_key=api_key)

model_dropdown = widgets.Dropdown(
    options=['gpt-3.5-turbo', 'gpt-4', 'gpt-4-turbo'],
    value='gpt-3.5-turbo',
    description='Modelo:'
)

prompt_textbox = widgets.Textarea(
    value='',
    placeholder='Digite o prompt aqui...',
    description='Prompt:',
    layout=widgets.Layout(width='100%', height='100px')
)

temperature_slider = widgets.FloatSlider(
    value=0.7,
    min=0.0,
    max=1.7,
    step=0.1,
    description='Temperature:'
)

max_tokens_dropdown = widgets.Dropdown(
    options=[50, 100, 150, 200, 250, 300, 400, 500],
    value=300,
    description='Max Tokens:'
)

top_p_dropdown = widgets.Dropdown(
    options=[0.1, 0.3, 0.5, 0.7, 0.9, 1.0],
    value=1.0,
    description='Top P:'
)

frequency_penalty_slider = widgets.FloatSlider(
    value=0.0,
    min=-2.0,
    max=2.0,
    step=0.1,
    description='Frequency Penalty:'
)

presence_penalty_slider = widgets.FloatSlider(
    value=0.0,
    min=-2.0,
    max=2.0,
    step=0.1,
    description='Presence Penalty:'
)

best_of_dropdown = widgets.Dropdown(
    options=[1, 2, 3, 4, 5],
    value=1,
    description='Best Of:'
)

run_button = widgets.Button(
    description='Executar',
    button_style='success'
)

output_area = widgets.Output()

@output_area.capture(clear_output=True)
def execute_openai(_):
      response = client.chat.completions.create(
          model=model_dropdown.value,
          messages=[
              {"role": "user", "content": prompt_textbox.value}
          ],
          temperature=temperature_slider.value,
          max_tokens=max_tokens_dropdown.value,
          top_p=top_p_dropdown.value,
          frequency_penalty=frequency_penalty_slider.value,
          presence_penalty=presence_penalty_slider.value,
          n=best_of_dropdown.value
      )
      print("\nResposta:\n")
      print(response.choices[0].message.content.strip())


run_button.on_click(execute_openai)

display(
    widgets.VBox([
        model_dropdown,
        prompt_textbox,
        temperature_slider,
        max_tokens_dropdown,
        top_p_dropdown,
        frequency_penalty_slider,
        presence_penalty_slider,
        best_of_dropdown,
        run_button,
        output_area
    ])
)
```
