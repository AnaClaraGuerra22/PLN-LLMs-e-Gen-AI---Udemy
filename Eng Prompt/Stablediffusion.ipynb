```python
from diffusers import DiffusionPipeline
import torch
import matplotlib.pyplot as plt
from PIL import Image
```

```python
pretrained_model_or_path ="stabilityai/stable-diffusion-xl-base-1.0"
pipeline = DiffusionPipeline.from_pretrained(pretrained_model_or_path,
                                         torch_dtype=torch.float16, use_safetensors=True).to("cuda")
pipeline.enable_attention_slicing()
```

```python
imagem = pipeline(
    prompt="Uma imagem realística de um peixe palhaço em um aquário",
    negative_prompt="Sem cavalos marinhos",
    num_inference_steps=75,
    height=768,
    width=768,
    guidance_scale=12
).images[0]
```

```python
plt.axis('off')
plt.imshow(imagem)
plt.show()
```
